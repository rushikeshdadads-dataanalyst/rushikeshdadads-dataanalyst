```python
import pandas as pd\nimport numpy as np\nimport matplotlib.pyplot as plt\nimport seaborn as sns
```

```python
df = pd.read_csv("../data/Geldium_PowerBI_Data.csv")\ndf.head()\n
```

```python
print("Shape:", df.shape)\ndisplay(df.info())\ndisplay(df.isna().sum().sort_values(ascending=False))\nprint("Duplicate Customer IDs:", df["Customer_ID"].duplicated().sum())
```

## Target distribution

```python
target = df["Delinquent_Account"].value_counts()\ndisplay(target)\nprint("Delinquency rate:", round(df["Delinquent_Account"].mean()*100,2), "%")
```

## Segment analysis

```python
for c in ["Age_Group","Credit_Score_Band","Employment_Status","Credit_Card_Type","Location"]:\n    out = df.groupby(c, dropna=False)["Delinquent_Account"].agg(["count","sum","mean"]).reset_index()\n    out["mean"] *= 100\n    print("\\n", c)\n    display(out.sort_values("mean", ascending=False))
```

## Numeric relationships

```python
num_cols = ["Age","Income","Credit_Score","Credit_Utilization","Missed_Payments","Loan_Balance","Debt_to_Income_Ratio","Account_Tenure","Delinquent_Account"]\ndisplay(df[num_cols].corr()["Delinquent_Account"].sort_values())
```

## Risk flag diagnostic

```python
tp = ((df.High_Risk_Flag==1)&(df.Delinquent_Account==1)).sum()\nflagged = (df.High_Risk_Flag==1).sum()\nactual = (df.Delinquent_Account==1).sum()\nprint("True positives:", tp)\nprint("Precision:", round(tp/flagged*100,2), "%")\nprint("Recall:", round(tp/actual*100,2), "%")
```

## Payment behaviour

```python
months = [f"Month_{i}" for i in range(1,7)]\nstatus_share = pd.DataFrame({m: df[m].value_counts(normalize=True)*100 for m in months}).fillna(0)\ndisplay(status_share)
```

## Conclusion\nThe data suggests that delinquency should be treated as a multi-factor risk problem. Segment-level patterns are useful for collections prioritization, but causal claims and automated decisions require further validation.
